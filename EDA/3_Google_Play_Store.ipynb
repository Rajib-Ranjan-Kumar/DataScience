```python
def intok(value):
    if isinstance(value, str):  # ensure it's a string
        value = value.replace('+', '').strip()

        if 'M' in value:
            value = value.replace('M', '')
            num = float(value) * 1_000_000
            return int(num)
        elif 'k' in value or 'K' in value:
            value = value.replace('k', '').replace('K', '')
            num = float(value) * 1_000
            return int(num)
        else:
            try:
                return int(value)
            except ValueError:
                return None
    return value
```

```python
df['Size_in_k'] = df['Size'].map(intok)

df.head()
```
