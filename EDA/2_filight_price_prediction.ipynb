```python
def intoMinute(time):
    time = str(time).strip()
    hour = 0
    minute = 0

    if 'h' in time:
        hour_part = time.split('h')[0]
        hour = int(hour_part.strip())

    if 'm' in time:
        minute_part = time.split('h')[-1].replace('m', '').strip()
        if minute_part:
            minute = int(minute_part)

    return hour * 60 + minute



```

```python
dfc['total_duration'] = dfc['Duration'].map(intoMinute)
dfc.head()

```
