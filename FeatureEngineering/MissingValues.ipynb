Mean


```python
df['mean_age']=df['age'].fillna(df['age'].mean())
df
```

```python
sns.histplot(df['mean_age'],kde=True)
```

Mean imputation works well when the data is normally distributed without outliers. However, if the data is skewed or contains outliers, median imputation is often a better choice as it is less affected by extreme values.

Median
Median imputation is useful when the data is skewed or contains outliers, as it provides a more robust measure of central tendency compared to the mean.

```python
df['median_age']=df['age'].fillna(df['age'].median())
df
```

```python
sns.histplot(df['median_age'],kde=True)
```

Mode Imputation technique is used to fill missing values in categorical data. The mode is the most frequently occurring value in a dataset. By replacing missing values with the mode, we can maintain the distribution of the categorical variable without introducing bias.

```python
df.isna().sum()
```

```python
df[df['embarked'].isnull()]
```

```python
df['embarked_notna']=df['embarked'].fillna(df['embarked'].mode())
df
```
