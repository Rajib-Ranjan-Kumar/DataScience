Label encoding is a technique used in machine learning and data preprocessing to convert categorical labels into a numerical format. This is particularly useful when working with algorithms that require numerical input. Each unique category is assigned a unique integer value.

```python
import pandas as pd
```

```python
df=pd.DataFrame({
    "color" : ['red','blue','green','blue']
})
df
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
encoder=LabelEncoder()
encoded=encoder.fit_transform(df[['color']])
encoded
```

```python
encoder.transform([['green']])
```

# Ordinal Encoding

ordinal encoding is a method used to convert categorical variables with a meaningful order or ranking into numerical format. In ordinal encoding, each category is assigned an integer value based on its order or rank. This technique is useful when the categories have a natural hierarchy, such as "low," "medium," and "high." eg: 'cold', 'warm', 'hot' 

```python
df=pd.DataFrame({
    'education':['+2','master','schooling','bachelor']
})
df
```

```python
from sklearn.preprocessing import OrdinalEncoder

encoder=OrdinalEncoder(categories=[['schooling','+2','bachelor','master']])
encoded=encoder.fit_transform(df[['education']])
encoded=pd.DataFrame(encoded,columns=encoder.get_feature_names_out())
encoded

```

```python
pd.concat([df,encoded],axis=1)
```
