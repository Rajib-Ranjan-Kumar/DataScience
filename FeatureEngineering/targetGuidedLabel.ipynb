```python
import seaborn as sns
import pandas as pd

# Load dataset
tips = sns.load_dataset('tips')

# Compute mean total_bill for each time category
encoding_map = tips.groupby('time')['total_bill'].mean().sort_values().to_dict()

# Apply target guided ordinal encoding
tips['time_encoded'] = tips['time'].map(encoding_map)


```
