```python
from sklearn.preprocessing import StandardScaler
ss=StandardScaler()
```

Dimension reduction technique which is used to reduce the number of features in a dataset by transforming the original features into a new set of features called principal components. These principal components are linear combinations of the original features and are ordered in such a way that the first few components capture most of the variance in the data.|

```python
from sklearn.decomposition import PCA
pca=PCA(n_components=3)
```

```python
fitting=pca.fit(s_data)
```

```python
three_feature=pca.fit_transform(s_data)
three_feature
```

```python
fitting.explained_variance_
```

```python
import plotly.express as px
import pandas as pd

# convert numpy array → DataFrame
df = pd.DataFrame(
    three_feature,
    columns=["pca1", "pca2", "pca3"]
)
y=dataset['target']
df["label"] = y.astype(str)

fig = px.scatter_3d(
    df,
    x="pca1",
    y="pca2",
    z="pca3",
    color="label",
    title="3D PCA Plot"
)
fig.show()
```
