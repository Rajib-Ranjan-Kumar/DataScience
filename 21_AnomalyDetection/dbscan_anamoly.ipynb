```python
import pandas as pd
import numpy as np
from sklearn.datasets import make_circles
```

```python
x,y=make_circles(n_samples=2000,noise=0.1,factor=0.3)
```

```python
y
```

```python
x
```

```python
import matplotlib.pyplot as plt
plt.scatter(x[:,0],x[:,1],c=y)
```

```python
from sklearn.cluster import DBSCAN
dbscan=DBSCAN(eps=0.10)
yp=dbscan.fit_predict(x)
```

```python
yp
```

```python
import matplotlib.pyplot as plt
plt.scatter(x[:,0],x[:,1],c=yp)
```
