```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
from sklearn.datasets import load_iris
```

```python
from sklearn.model_selection import train_test_split
```

```python
x,y=load_iris(return_X_y=True)
```

```python
x
```

```python
y
```

```python
xtrain,xtest,ytrain,ytest=train_test_split(x,y,test_size=0.25,random_state=10)
```

```python
from sklearn.naive_bayes import GaussianNB
gnb=GaussianNB()
```

```python
gnb.fit(xtrain,ytrain)
```

```python
y_pred=gnb.predict(xtest)
```

```python
from sklearn.metrics import confusion_matrix,classification_report,accuracy_score
print(confusion_matrix(y_pred,ytest))
print(accuracy_score(y_pred,ytest))
print(classification_report(y_pred,ytest))
```

```python
from sklearn.preprocessing import OneHotEncoder
one=OneHotEncoder()
array=one.fit_transform(df[['day']]).toarray()
df2=pd.DataFrame(data=array,columns=one.get_feature_names_out())
df2.head()
```

```python
df=pd.concat([df,df2],axis=1)
df.drop(columns='day',axis=-1,inplace=True)
```

```python
df.head()
```

```python
#independent and dependent
y=df['time']
x=df.drop(columns='time',axis=-1)
```

```python
y
```

```python
x
```

```python
xtrain,xtest,ytrain,ytest=train_test_split(x,y,test_size=0.25,random_state=42)
```

```python
from sklearn.naive_bayes import BernoulliNB
brn=BernoulliNB()
brn.fit(xtrain,ytrain)
```

```python
y_pred=brn.predict(xtest)
```

```python
from sklearn.metrics import confusion_matrix,classification_report,accuracy_score
print(confusion_matrix(y_pred,ytest))
print(accuracy_score(y_pred,ytest))
print(classification_report(y_pred,ytest))
```
