```python
#independent and dependent feature
x=df[['tip', 'sex', 'smoker', 'day', 'time', 'size']]
y=df[['total_bill']]
```

```python
#train test split
from sklearn.model_selection import train_test_split
xtrain,xtest,ytrain,ytest=train_test_split(x,y,test_size=0.3,random_state=10)
```

```python
from sklearn.preprocessing import LabelEncoder
le1=LabelEncoder()
le2=LabelEncoder()
le3=LabelEncoder()
xtrain['sex']=le1.fit_transform(xtrain['sex'])
xtrain['smoker']=le2.fit_transform(xtrain['smoker'])
xtrain['time']=le3.fit_transform(xtrain['time'])
```

```python
from sklearn.preprocessing import LabelEncoder
xtest['sex']=le1.transform(xtest['sex'])
xtest['smoker']=le2.transform(xtest['smoker'])
xtest['time']=le3.transform(xtest['time'])
```

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
encoder=OneHotEncoder()
e=encoder.fit_transform(xtrain[['day']]).toarray()
e_df=pd.DataFrame(e,columns=encoder.get_feature_names_out())
e_df.head()
```

```python
encoder = OneHotEncoder()
e = encoder.fit_transform(xtrain[['day']]).toarray()
e_df = pd.DataFrame(e, columns=encoder.get_feature_names_out(), index=xtrain.index)

# correct concatenation (keep same row index)
xtrain = pd.concat([xtrain, e_df], axis=1)

xtrain.head()
```

```python
encoder = OneHotEncoder()
e = encoder.fit_transform(xtest[['day']]).toarray()
e_df = pd.DataFrame(e, columns=encoder.get_feature_names_out(), index=xtest.index)

# correct concatenation (keep same row index)
xtest = pd.concat([xtest, e_df], axis=1)

xtest.head()
```

```python
xtest.drop(columns='day',axis=-1,inplace=True)
xtest.head()
```

```python
xtrain.drop(columns='day',axis=-1,inplace=True)
xtrain.head()
```

```python
from sklearn.svm import SVR
s=SVR()
s.fit(xtrain,ytrain)
```

```python
y_pred=s.predict(xtest)
```

```python
from sklearn.metrics import r2_score,mean_absolute_error
print(r2_score(ytest,y_pred))
print(mean_absolute_error(ytest,y_pred))
```
