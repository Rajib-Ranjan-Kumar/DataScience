```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
import seaborn as sns
```

```python
from sklearn.datasets import fetch_california_housing
```

```python
dataset=fetch_california_housing()
```

```python
dataset
```

```python
dataset.keys()
```

```python
df=pd.DataFrame(data=dataset.data,columns=dataset.feature_names)
```

```python
df.head()
```

```python
df['price']=dataset.target
df.head()
```

```python
df.info()
```

```python
df.isna().sum()
```

```python
plt.figure(figsize=(15,15))
sns.heatmap(df.corr(),annot=True)

plt.show()
```

```python
x=df.iloc[:,:-1]
y=df.iloc[:,-1:]
x.head()
```

```python
y
```

```python
from sklearn.model_selection import train_test_split
xtrain,xtest,ytrain,ytest=train_test_split(x,y,random_state=42,train_size=0.33)
xtrain
```

```python
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
xtrain=scaler.fit_transform(xtrain)
xtest=scaler.transform(xtest)
```

```python
from sklearn.linear_model import LinearRegression
regression=LinearRegression(

)
regression.fit(xtrain,ytrain)
```

```python
y_predict=regression.predict(xtest)

regression.coef_
```

```python
regression.intercept_
```

```python
# assumptions
plt.scatter(ytest,y_predict)
plt.show()
```

```python
residuals=ytest-y_predict
residuals
```

```python
sns.displot(residuals,kind='kde')
plt.show()
```

```python
plt.scatter(y_predict,residuals)
plt.show()
```

Pickling in Python is the process of converting a Python object into a byte stream, which can then be saved to a file or transmitted over a network. This process is known as serialization. The `pickle` module in Python provides functions to perform pickling and unpickling (the reverse process of converting a byte stream back into a Python object).

```python
import pickle
pickle.dump(regression,open('regression.pkl','wb'))
```

```python
model=pickle.load(open('regression.pkl','rb'))
```

```python
model.predict(xtest)
```
