```python
df=pd.read_csv('batter.csv')
df
```

```python


def choose(val):
    if val < 20:
        return 'red'
    elif val<30:
        return 'blue'
    else:
        return 'green'
    
plt.scatter(df['avg'], df['strike_rate'], color=df['avg'].apply(choose), marker="*")
plt.title("Average vs Strike Rate")
plt.xlabel("Average")
plt.ylabel("Strike Rate")
plt.legend()
plt.show()
```
