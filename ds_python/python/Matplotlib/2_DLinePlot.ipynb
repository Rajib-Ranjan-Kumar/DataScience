```python
import numpy as numpy
import pandas as pd
import matplotlib.pyplot as plt

```

```python
df=pd.read_csv('sharma-kohli.csv')
df
```

```python
plt.plot(df['index'],df['V Kohli'])
plt.plot(df['index'],df['RG Sharma'])
```

### providing label of axis

```python
plt.plot(df['index'],df['V Kohli'])
plt.plot(df['index'],df['RG Sharma'])
plt.xlabel("Seasons of IPL")
plt.ylabel("Runs Scored")
plt.title("Comparison between Rohit Sharma and Virat Kholi")
```

```python
plt.plot(df['index'],df['V Kohli'],color="Red")
plt.plot(df['index'],df['RG Sharma'],color="blue")
plt.xlabel("Seasons of IPL")
plt.ylabel("Runs Scored")
plt.title("Comparison between Rohit Sharma and Virat Kholi")
```

```python
plt.plot(df['index'],df['V Kohli'],color="Red",linestyle='dotted')
plt.plot(df['index'],df['RG Sharma'],color="blue",linestyle="dashdot")
plt.xlabel("Seasons of IPL")
plt.ylabel("Runs Scored")
plt.title("Comparison between Rohit Sharma and Virat Kholi")
```

### marker


```python
plt.plot(df['index'],df['V Kohli'],color="Red",linestyle='dotted',marker=".",markersize=10)
plt.plot(df['index'],df['RG Sharma'],color="blue",linestyle="dashdot",marker='*',markersize=8)
plt.xlabel("Seasons of IPL")
plt.ylabel("Runs Scored")
plt.title("Comparison between Rohit Sharma and Virat Kholi")
```

```python
plt.plot(df['index'],df['V Kohli'],color="Red",linestyle='dotted',marker=".",markersize=10,label="Virat")
plt.plot(df['index'],df['RG Sharma'],color="blue",linestyle="dashdot",marker='*',markersize=8,label="Rohit")
plt.xlabel("Seasons of IPL")
plt.ylabel("Runs Scored")
plt.title("Comparison between Rohit Sharma and Virat Kholi")
plt.legend(loc='upper left')
plt.grid()
```
