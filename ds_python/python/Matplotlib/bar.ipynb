```python
df=pd.read_csv('batsman_season_record.csv')
df
```

```python
x=np.arange(df.shape[0])
width=0.2
plt.bar(x-width,df['2015'],width,color='orange',label="2015")
plt.bar(x,df['2016'],width,color='skyblue',label="2016")
plt.bar(x+width,df['2017'],width,color='lightgreen',label="2017")
plt.xticks(x,df['batsman'],rotation=70)
plt.xlabel("2015,2016,2017 Ipl seasons")
plt.ylabel("Runs")
plt.legend()

```

Stack bar chart

```python
plt.bar(df['batsman'],df['2015'],label='2015')
plt.bar(df['batsman'],df['2016'],bottom=df['2015'],label='2016')
plt.bar(df['batsman'],df['2017'],bottom=df['2015']+df['2016'],label='2017')
plt.legend()
```
