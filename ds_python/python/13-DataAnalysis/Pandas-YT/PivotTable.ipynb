```python
import pandas as pd
import numpy as np

data = {
    'Date': pd.date_range('2023-01-01', periods=20),
    'Product': ['A', 'B', 'C', 'D']*5,
    'Region': ['East', 'West', 'North', 'South']*5,
    'Sales': np.random.randint(100, 1000, 20),
    'Units': np.random.randint(10, 100, 20),
    'Rep': ['John', 'Mary', 'Bob', 'Alice']*5
}

df = pd.DataFrame(data)
df
```

```python
pd.pivot_table(data=df,values='Sales',index='Region',columns='Product')
```

```python
pd.pivot_table(data=df,values=['Sales','Units'],index='Region',columns='Product')
```

### cross tabs

```python
pd.crosstab(index=df['Region'],columns=df['Product'])
```
