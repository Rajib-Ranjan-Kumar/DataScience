**Merging**

```python
import pandas as pd

df1 = pd.DataFrame({
    'EmployeeID': [101, 102, 103, 104],
    'Name': ['Alice', 'Bob', 'Charlie', 'David'],
    'Department': ['HR', 'Finance', 'IT', 'Marketing']
})


df2 = pd.DataFrame({
    'EmployeeID': [102, 103, 104, 105],
    'Salary': [50000, 60000, 55000, 45000]
})
print(df1)
print(df2)
```

**default Merge**

```python
pd.merge(df1,df2)
```

**Merge with on**



```python
pd.merge(df1,df2,on='EmployeeID')
```

**Merge with how**

```python
pd.merge(df1,df2,how='outer')
```

```python
pd.merge(df1,df2,how='left')
```

```python
pd.merge(df1,df2,how='right')
```

## concatenate

```python
pd.concat([df1,df2])
```

```python
pd.concat([df1,df2],axis=1)
```

## Joining

```python

df1 = pd.DataFrame({
    'Name': ['Alice', 'Bob', 'Charlie', 'David'],
    'Department': ['HR', 'Finance', 'IT', 'Marketing']
}, index=[101, 102, 103, 104])


df2 = pd.DataFrame({
    'Salary': [50000, 60000, 55000, 45000]
}, index=[102, 103, 104, 105])

```

```python
df1
```

```python
df2
```

```python
df1.join(df2)
```

```python
df2.join(df1)
```

```python
df1.join(df2,how='outer')
```

```python
df2.join(df1,how='right')
```
