```python
import numpy as np
import pandas as pd

data = {
    'A': [1,2, np.nan,4,5],
    'B': [np.nan,2,3,4,5],
    'C': [1,2,3,np.nan,np.nan],
    'D': [1,np.nan,np.nan,np.nan,5]
}

df = pd.DataFrame(data)
df
```

```python
df.isna()
```

```python
df.isna().sum()
```

```python
df.isna().any()
```

## Removing Missing DAta


```python
df.dropna()
```

```python
df.dropna(thresh=3)
```

## Filling Missing Data

```python
df.fillna(0)
```

```python
values={'A':0,'B':100,'C':200,'D':400}
df.fillna(value=values)
```

```python
values={'A':df['A'].mean(),'B':df['B'].mean(),'C':df['C'].mean(),'D':df['D'].mean()}
df.fillna(value=values)
```
