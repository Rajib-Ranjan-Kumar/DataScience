```python
import numpy as np

#creating array from list
arr1=[1,2,3,45]
print(arr1)


#creating array from the numpy
arr=np.array([1,2,3,3])
print(arr)

#creating 2darray from the numpy
arr=np.array([[1,2,3,3],[4,5,6,7]])
print(arr)
```

```python
#multiplying by 2
#list
arr1=arr1*2
print(arr1)#increase each element make double size

arr=arr*2
print(arr)#multiplying each element by 2
```

```python
zeroMatrix=np.zeros((2,3))
print(zeroMatrix)

oneMatrix=np.ones((2,3))
print(oneMatrix)

#identity matrix 
identiy=np.eye(3)
print(identiy)

specific=np.full((2,3),7)
print(specific)
```

```python
#random
random=np.random.random((2,3))
print(random)

sequence=np.arange(0,11,2)
print(sequence)
print(sequence.reshape(2,3))
```

```python
#Tensor in python is array with morethan 2d

tensor=np.array(
    [
        [
            [1,2,3],[3,2,1]
        ],
        [
            [6,7,8],[8,7,6]
        ]
    ])
print(tensor)
```

```python
#Array properties
arr=np.array([[1,2,3],[3,4,5]]
            )
print(arr.ndim)
print(arr.dtype)
print(arr.shape)
print(arr.size)
flattened=arr.flatten()
print(flattened)
ravel=arr.ravel()
print(ravel)
Transpose=flattened.T
print(Transpose)
```
