# Advance operation with Business examples

```python
import numpy as np
import matplotlib.pyplot as plt

# Data structure: [restaurant_id, 2021, 2022, 2023, 2024]
sales_data = np.array([
[1, 150000,180000, 220000, 250000], # Paradise Biryani
[2, 120000, 140000, 160000, 190000], # Beijing Bites
[3, 200000, 230000, 260000, 300000], # Pizza Hub
[4, 180000, 210000, 240000, 270000], # Burger Point
[5, 160000, 185000, 205000, 230000], # Chai Point
])
```

```python
import numpy as np
#total sales per year
print(np.sum(sales_data,axis=0))
yearly_total=np.sum(sales_data[:,1:],axis=0)
print(yearly_total)
```

```python
#minimum sales per resturant
min_sale=np.min(sales_data[:,1:],axis=1)
print(min_sale)
#minimum sales per year
min_saley=np.min(sales_data[:,1:],axis=0)
print(min_saley)
```

```python
#maximum sales per year
max_sale=np.max(sales_data[:,1:],axis=1)
print(min_sale)
```

```python
#Average sales per restaurant
average=np.mean(sales_data[:,1:],axis=1)
print(average)
```

```python
import matplotlib.pyplot as plt
cumsum = np.cumsum(sales_data[:,1:],axis=1)
print(cumsum)

plt.figure(figsize=(5,5))
plt.plot(np.mean(cumsum,axis=0))
plt.title("Average Cumulative sales accross all restaurant")
plt.xlabel("Years")
plt.ylabel("Sales")
plt.grid(True)
plt.show
```

```python
vector1=np.array([1,2,3,4,5,6])
vector2=np.array([1,4,7,8,9,4])
print("Vector addition:",vector1+vector2)
print("Vector multiplication:",vector1*vector2)
print("Vector subtraction:",vector1-vector2)
print("Vector division:",vector1/vector2)
print("\n Dot product ",np.dot(vector1,vector2))
angle=np.arccos(np.dot(vector1,vector2)/(np.linalg.norm(vector1)*np.linalg.norm(vector2)))
print(angle)
```

```python
#vectorized operation
restaurant_types=np.array(["Biryani","Chinese"])
vectorized_upper=np.vectorize(str.upper)
upper=vectorized_upper(restaurant_types)
print(upper)
```
