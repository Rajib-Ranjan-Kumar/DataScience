```python
#importing numpy
import numpy as np

#creating arrays
arr= np.array([1,2,3,4,5])
print(arr)
arr
```

```python
print(type(arr))
```

```python
#shape
print(arr.shape)
```

```python
arr= np.array([1,2,3,4,5,6])
arr1=arr.reshape(2,3)
print(arr1)
print(arr1.shape)
```

```python
arr3=np.array([[1,2,3],[4,5,6]])
print(arr3)
print(arr3.shape)
```

```python
#arange,ones,eye
#arange
arr2=np.arange(0,10,2)#[0,2,4,6,8]
print(arr2)
arr_reshaped=arr2.reshape(5,1)
print(arr_reshaped)

#ones
arr3=np.ones((3,3))
print(arr3)

arr4=np.zeros((2,2,))
print(arr4)

#eye
arr5=np.eye(3)
print(arr5)
```

```python
#shape,ndim,size,dtype,itemsize
arr=np.array([1,2,3,4,5])
print(arr.shape)
print(arr.ndim)
print(arr.size)
print(arr.dtype)
print(arr.itemsize)
```

```python
#Vectorized operation
arr1=np.array([2,3,4,5])
arr2=np.array([4,3,4,6])
print(arr1+arr2)
print(arr1-arr2)
print(arr1*arr2)
print(arr1/arr2)
```

```python
#universal functions->sqrt,exp,sin and log
arr1=np.array([2,3,4,5])
print(np.sqrt(arr1))
print(np.exp(arr1))
print(np.sin(arr1))
print(np.log(arr1))
```

```python
arr1=np.array([[2,3,4,5],[3,4,5,6],[4,5,6,7]])
print(arr1)

print("\n",arr1[1:])
print("\n",arr1[0, 0])
print("\n",arr1[1:][1:3])
```

```python
#modifying array
arr1[0,0]=100
print(arr1)
arr1[1:]=100
print(arr1)
```

```python
#Statistical Operations and Normalization
data=[1,2,3,4,5,6,7,8]
mean=np.mean(data)

median=np.median(data)
std=np.std(data)
var=np.var(data)
print(np.mean(data))
print(median)
print(std)
print(var)
```
