```python
### Assignment 1: Creating and Connecting to a Database

# 1. Write a Python function to create a new SQLite3 database named `test.db`.
# 2. Write a Python function to create a table named `employees` with columns 
# `id` (integer), `name` (text), `age` (integer), and `department` (text) in the
#  `test.db` database.

import sqlite3
connection=sqlite3.connect('test.db')
cursor=connection.cursor()
cursor.execute('''
  Create Table if not exists employee(
               id integer primary key,
               name text not null,
               age integer,
               department text
               )
''')

```

```python
connection.close()
```

```python

# ### Assignment 2: Inserting Data

# 1. Write a Python function to insert a new employee into the `employees` table.
# 2. Insert at least 5 different employees into the `employees` table.
def Insert(cursor, id, name, age, department):
    try:
        cursor.execute('''
            INSERT INTO employee(id, name, age, department)
            VALUES (?, ?, ?, ?)
        ''', (id, name, age, department))
    except sqlite3.IntegrityError as e:
        print(f"Error inserting ID {id}: {e}")

# Insert employees
Insert(cursor, 4, "rajib", 20, "technology")
Insert(cursor, 2, "raj", 22, "technology")
Insert(cursor, 0, "ranjan", 21, "technology")
Insert(cursor, 5, "rabi", 24, "science")     # Changed ID from 4 to 5
Insert(cursor, 1, "rohit", 19, "technology")

# Commit the changes
connection.commit()

    
```
