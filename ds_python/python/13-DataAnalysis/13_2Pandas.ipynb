```python
import pandas as pd
list=[1,2,3,4]
series=pd.Series(list)
print(series)
```

```python
#create a series from dictionaries 
dic={"name":"RAjib","roll":23053748}
seriesd=pd.Series(dic)
print(seriesd)
```

```python
#Creating series from with indexes
indexes=[1,2,3]
data=["rajib","ranjan","jaiswal"]
seriesI=pd.Series(data=data,index=indexes)
print(seriesI)
```

```python
#DataFrame
#creating data from dictionary lists
d_list={
    "name":["rajib","ranjan","jaiswal"],
    "age":[23,21,22],
    "City":["Delhi","Mumbai","Ranchi"]
}
df = pd.DataFrame(d_list)
print(df)
```

```python
#creating dataframes from list of dictionaries
list = [
    {"name":"RAjib","age":45},
    {"name":"raj","age":43},
    {"name":"rabi","age":23}
        ]

df = pd.DataFrame(list)
print(df)
```
