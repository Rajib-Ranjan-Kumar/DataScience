```python
x=df.drop(columns=['selling_price'],axis=1)
y=df['selling_price']
```

```python
oneH_feature = ['transmission_type', 'fuel_type', 'seller_type']
label_feature = ['model']   # MUST be a list
stand_feature = x.select_dtypes(exclude="object").columns.tolist()
```

```python
from sklearn.model_selection import train_test_split
xtrain,xtest,ytrain,ytest=train_test_split(x,y,test_size=0.25,random_state=42)
```

```python
from sklearn.preprocessing import OneHotEncoder, OrdinalEncoder, StandardScaler
from sklearn.compose import ColumnTransformer

oneH_feature = ['transmission_type', 'fuel_type', 'seller_type']
label_feature = ['model']
stand_feature = x.select_dtypes(exclude='object').columns.tolist()

preprocessor = ColumnTransformer(
    transformers=[
        ("onehot", OneHotEncoder(drop='first', handle_unknown='ignore'), oneH_feature),
        ("ordinal", OrdinalEncoder(handle_unknown='use_encoded_value', unknown_value=-1), label_feature),
        ("scaler", StandardScaler(), stand_feature)
    ]
)
```

```python
type(xtrain)
```

```python
xtr = preprocessor.fit_transform(xtrain)
xts = preprocessor.transform(xtest)
```

```python
xtr
```

```python
onehot_cols = (
    preprocessor
    .named_transformers_['onehot']
    .get_feature_names_out(oneH_feature)
)
ordinal_cols = label_feature
scaled_cols = stand_feature
final_columns = list(onehot_cols) + ordinal_cols + scaled_cols
import pandas as pd

xtrain= pd.DataFrame(xtr, columns=final_columns, index=xtrain.index)
xtest= pd.DataFrame(xts, columns=final_columns, index=xtest.index)
```

```python
from sklearn.ensemble import RandomForestRegressor
rfr=RandomForestRegressor()
rfr.fit(xtrain,ytrain)
```

```python
y_pred=rfr.predict(xtest)
```

```python
from sklearn.metrics import mean_absolute_error,mean_squared_error,mean_absolute_percentage_error
print(mean_squared_error(y_pred,ytest))
print(mean_absolute_error(y_pred,ytest))
print(mean_absolute_percentage_error(y_pred,ytest))
```
