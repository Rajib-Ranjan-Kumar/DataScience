```python
df['TotalVisiting']=df['NumberOfPersonVisiting']+df['NumberOfChildrenVisiting']
df.drop(columns=['NumberOfPersonVisiting','NumberOfChildrenVisiting'],axis=1,inplace=True)
df.columns
```

```python
x=df.drop(columns=['ProdTaken'],axis=1)
y=df['ProdTaken']
```

```python
from sklearn.model_selection import train_test_split
xtrain,xtest,ytrain,ytest=train_test_split(x,y,test_size=0.25,random_state=42)
```

```python
xtrain
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder,StandardScaler
cat_feature=x.select_dtypes(include="object").columns
num_feature=x.select_dtypes(exclude='object').columns
num_transformer=StandardScaler()
oh_transformer=OneHotEncoder(drop='first')
```

```python
cat_feature
```

```python
preprocessor=ColumnTransformer([
    ("OneHotEncoder",oh_transformer,cat_feature),
    ("StandardScaler",num_transformer,num_feature)]
)
```

```python
xtrain=preprocessor.fit_transform(xtrain)
xtest=preprocessor.transform(xtest)
```

```python
xtrain
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score,classification_report,confusion_matrix,precision_score,recall_score,f1_score,roc_auc_score,roc_curve
```

```python
models={
    "Random Forest":RandomForestClassifier(n_estimators= 100, min_samples_split= 2, min_samples_leaf= 1, max_features='sqrt', max_depth= 30)
}
for i in range(len(list(models))):
    model=list(models.values())[i]
    model.fit(xtrain,ytrain)
    ytrain_pred=model.predict(xtrain)
    ytest_pred=model.predict(xtest)

    modeltrain_accuracy=accuracy_score(ytrain_pred,ytrain)
    modeltrain_f1=f1_score(ytrain,ytrain_pred)
    modeltrain_precision=precision_score(ytrain,ytrain_pred)
    modeltrain_recall=recall_score(ytrain,ytrain_pred)
    modeltrain_rocauc_score=roc_auc_score(ytrain,ytrain_pred)


    modeltest_accuracy=accuracy_score(ytest_pred,ytest)
    modeltest_f1=f1_score(ytest,ytest_pred)
    modeltest_precision=precision_score(ytest,ytest_pred)
    modeltest_recall=recall_score(ytest,ytest_pred)
    modeltest_rocauc_score=roc_auc_score(ytest,ytest_pred)

    print(list(models.keys())[i])

    print('Model performance for Training set')
    print('- Accuracy: {:.4f}'.format(modeltrain_accuracy))
    print('- F1 score: {:.4f}'.format(modeltrain_f1))

    print('- Precision: {:.4f}'.format(modeltrain_precision))
    print('- Recall: {:.4f}'.format(modeltrain_recall))
    print('- Roc Auc Score: {:.4f}'.format(modeltrain_rocauc_score))

    print('-----------------------------------')

    print('Model performance for Test set')
    print('- Accuracy: {:.4f}'.format(modeltest_accuracy))
    print('- F1 score: {:.4f}'.format(modeltest_f1))
    print('- Precision: {:.4f}'.format(modeltest_precision))
    print('- Recall: {:.4f}'.format(modeltest_recall))
    print('- Roc Auc Score: {:.4f}'.format(modeltest_rocauc_score))

    print('='*35)
    print('\n')
```

```python
#Hyper parameter tunning
param_rf = {
    'n_estimators': [100, 200, 500],
    'max_depth': [10, 20, 30, None],
    'min_samples_split': [2, 5, 10],
    'min_samples_leaf': [1, 2, 4],
    'max_features': ['sqrt', 'log2']
}
```

```python
from sklearn.model_selection import RandomizedSearchCV

models_rcv=[
    ("rf" ,RandomForestClassifier(),param_rf)
]
```

```python
from sklearn.model_selection import RandomizedSearchCV

# 1. Initialize the dictionary OUTSIDE the loop
model_results = {}

for name, model, params in models_rcv:
    print(f"Starting tuning for: {name}...")
    
    random = RandomizedSearchCV(
        estimator=model,
        param_distributions=params,
        cv=3,
        verbose=1, # Reduced verbose slightly to keep console cleaner
        n_iter=100,
        n_jobs=-1,
        random_state=42 # Added for reproducibility
    )
    
    random.fit(xtrain, ytrain)
    
    # 2. Store more than just params (Best Params + Best Score)
    model_results[name] = {
        'params': random.best_params_,
        'score': random.best_score_,
        'best_model': random.best_estimator_
    }

# 3. Printing the results
for model_name, results in model_results.items():
    print(f"\n{'='*20} {model_name} {'='*20}")
    print(f"Best Score: {results['score']:.4f}")
    print(f"Best Params: {results['params']}")
```
