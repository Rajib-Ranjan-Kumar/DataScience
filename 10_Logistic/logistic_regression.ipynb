```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```python
from sklearn.datasets import make_classification
x,y=make_classification(n_samples=1000,n_features=10,n_classes=2,random_state=42)
```

```python
x=pd.DataFrame(x)
y=pd.DataFrame(y)
x.head()
```

```python
from sklearn.model_selection import train_test_split
xtrain,xtest,ytrain,ytest=train_test_split(x,y,test_size=0.30,random_state=42)
```

```python
xtrain
```

```python
from sklearn.linear_model import LogisticRegression
logistic=LogisticRegression()
```

```python
logistic.fit(xtrain,ytrain)
```

```python
y_pred=logistic.predict(xtest)
y_pred
```

```python
logistic.predict_proba(xtest)
```

```python
from sklearn.metrics import accuracy_score,precision_score,confusion_matrix,recall_score,f1_score,classification_report
print("accuracy_score:",accuracy_score(ytest,y_pred))
print("accuracy_score:",precision_score(ytest,y_pred))
print("accuracy_score:",recall_score(ytest,y_pred))
print("accuracy_score:",f1_score(ytest,y_pred))
print("accuracy_score:",confusion_matrix(ytest,y_pred))
print(classification_report(ytest,y_pred))
```

```python
cm=confusion_matrix(ytest,y_pred)
sns.heatmap(cm,fmt="d",cmap="Blues",annot=True)
plt.xlabel('predicted')
plt.ylabel('actual')
plt.show()
```
